```python
import numpy as np
import pandas as pd
from sklearn.ensemble import RandomForestRegressor

# ==========================================
# 1. 설정 및 CSV 데이터 로드
# ==========================================
CSV_FILENAME = "california_housing.csv"
FEATURES = ["MedInc", "HouseAge", "AveRooms", "Population", "AveOccup", "Latitude", "Longitude"]
TARGET = "MedHouseVal"
n_estimators=100
random_state=42
n_jobs=-1

# 데이터 불러오기
df = pd.read_csv(CSV_FILENAME)
X = df[FEATURES].values
y = df[TARGET].values

# ==========================================
# 2. 랜덤 포레스트 모델 선언 및 학습
# ==========================================
print("\n=== 랜덤 포레스트 모델 학습 시작 ===")

# n_estimators: 결정 트리 개수, n_jobs=-1: 모든 CPU 코어 사용
model = RandomForestRegressor(n_estimators, random_state, n_jobs)
model.fit(X, y)

print("\n 파이프라인 연동 테스트 완료!")

# ==========================================
# 3. 무작위 추출 10개 샘플 평가
# ==========================================
# 전체 데이터셋 크기 중에서 무작위로 10개 인덱스 선택
random_indices = np.random.choice(len(df), size=10, replace=False)

# 선택된 무작위 인덱스의 데이터 추출
sample_x_subset = X[random_indices]
sample_y_subset = y[random_indices]

# 추론 실행
preds = model.predict(sample_x_subset)

# 실제값 및 예측값 정리
actuals = sample_y_subset.flatten()
predictions = preds.flatten()

# 절대 오차 및 오차율(%) 연산: |실제값 - 예측값| / 실제값 * 100
abs_error = np.abs(actuals - predictions)
error_rate_pct = (abs_error / np.maximum(actuals, 1e-8)) * 100

# 보기 쉽게 Pandas DataFrame으로 정리
result_df = pd.DataFrame({
    "원본 Index": random_indices,  # 몇 번째 행 데이터를 뽑았는지 확인용
    "실제값 (Actual)": np.round(actuals, 4),
    "예측값 (Pred)": np.round(predictions, 4),
    "절대 오차": np.round(abs_error, 4),
    "오차율 (%)": np.round(error_rate_pct, 2)
})

print("\n=== 📊 무작위 추출 10개 샘플 예측 결과 비교 ===")
print(result_df.to_string(index=False))
```
